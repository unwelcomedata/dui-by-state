# Data Exploration

Scratch notebook for querying `project.duckdb` directly.  
Not part of the ingest → clean → prepare pipeline.

```python
import duckdb
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 60)
pd.set_option('display.width', 200)

con = duckdb.connect('../data/project.duckdb', read_only=True)

def q(sql):
    """Run SQL, return DataFrame."""
    return con.sql(sql).df()

print('Connected. Tables:')
q('SHOW TABLES')
```

```python
# What columns does the master states table have?
q('DESCRIBE states')
```

```python
# Quick look at the master table
q('SELECT * FROM states LIMIT 5')
```

```python
# Write your queries below — just use q('...')
q('''
SELECT state_abbr, pop_2024, pct_impaired_fars_2024
FROM states
ORDER BY pct_impaired_fars_2024 DESC
LIMIT 10
''')
```
